## ¿Cómo funciona la "Magia" Genética?

En lugar de calcular la solución perfecta paso a paso, los Algoritmos Genéticos imitan la selección natural de Charles Darwin:

La Población Inicial (Mutantes y Exploradores): Creamos al azar muchas mochilas diferentes (nuestra "población" de posibles soluciones).

La Prueba de Supervivencia (Fitness): Evaluamos qué tan buena es cada mochila (cuánto valor acumula sin pasarse del peso). Las mejores "sobreviven".

Reproducción (Cruce / Crossover): Tomamos a las mejores mochilas y "cruzamos" sus contenidos para crear nuevas mochilas "hijas" que heredan lo mejor de ambas.

Mutación (Toque de Azar): De vez en cuando, cambiamos un objeto al azar para explorar nuevas combinaciones y evitar quedarnos atrapados en malas ideas.

Repetimos este ciclo durante cientos de generaciones... ¡hasta que la evolución encuentra la combinación perfecta (o muy cercana a la ideal)!
