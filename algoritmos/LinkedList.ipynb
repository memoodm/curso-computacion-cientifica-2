# Linked List : Custom

```python
class Node:
  def __init__(self, valor, siguiente):
    self.valor = valor
    self.siguiente = siguiente
```

```python
class LinkedList:

    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def append(self, elemento):
        new_node = Node(elemento, None)
        if self.size == 0:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.siguiente = new_node
            self.tail = new_node
        self.size += 1

    def insert(self, indice, elemento):
        if indice < 0 or indice > self.size:
            raise IndexError("Índice fuera de rango")
        elif indice == 0:
            new_node = Node(elemento, self.head)
            self.head = new_node
            if self.size == 0:
                self.tail = new_node
            self.size += 1
        elif indice == self.size:
            self.append(elemento)
        else:
          actual = self.head
          for _ in range(indice - 1):
              actual = actual.siguiente

          new_node = Node(elemento, actual.siguiente)
          actual.siguiente = new_node
          self.size += 1

    def remove(self, elemento):
        if self.size == 0:
            return False
        actual = self.head
        anterior = None
        while actual is not None:
            if actual.valor == elemento:
                if anterior is None:
                    self.head = actual.siguiente
                    if self.size == 1:
                        self.tail = None
                else:
                    anterior.siguiente = actual.siguiente
                    if actual == self.tail:
                        self.tail = anterior
                self.size -= 1
                return True

            anterior = actual
            actual = actual.siguiente
        return False

    def __getitem__(self, indice):
        if indice < 0 or indice >= self.size:
            raise IndexError("Índice fuera de rango")
        actual = self.head
        for _ in range(indice):
            actual = actual.siguiente
        return actual.valor

    def __setitem__(self, indice, elemento):
        if indice < 0 or indice >= self.size:
            raise IndexError("Índice fuera de rango")
        actual = self.head
        for _ in range(indice):
            actual = actual.siguiente
        actual.valor = elemento

    def __iter__(self):
        actual = self.head
        while actual is not None:
            yield actual.valor
            actual = actual.siguiente

    def __len__(self):
        return self.size

    def __str__(self):
        lista = [None]*self.size
        actual_node = self.head
        index = 0
        while actual_node is not None:
            lista[index] = actual_node.valor
            actual_node = actual_node.siguiente
            index += 1
        return str(lista)

    def __repr__(self):
        return self.__str__()

    def __contains__(self, elemento):
        actual = self.head
        while actual is not None:
            if actual.valor == elemento:
                return True
            actual = actual.siguiente
        return False
```
