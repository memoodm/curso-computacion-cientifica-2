```python
class OrderList:

    def __init__(self):
        self.memory = []

    def add(self, element):
      """Inserta un elemento manteniendo el orden ascendente mediante búsqueda binaria."""
      low = 0
      high = len(self.memory)
      # Búsqueda binaria para encontrar la posición correcta (O(log n))
      while low < high:
          mid = (low + high) // 2
          if self.memory[mid] < element:
              low = mid + 1
          else:
              high = mid
      # Inserción en la posición encontrada
      self.memory.insert(low, element)

    def remove(self):
        if self.is_empty():
            raise IndexError("pop from empty orderlist")
        return self.memory.pop(0)

    def peek(self):
        if self.is_empty():
            raise IndexError("peek from empty orderlist")
        return self.memory[0]

    def is_empty(self):
        return not self.memory

    def __len__(self):
        return len(self.memory)

    def __str__(self):
        return str(self.memory)

    def __repr__(self):
        return f"OrderList({self.memory})"

    def __eq__(self, other):
        if isinstance(other, Stack):
            return self.memory == other.memory
        return False

    def __getitem__(self, index):
        return self.memory[index]

    def __contains__(self, element):
        return element in self.memory

    def __iter__(self):
        for e in self.memory:
            yield e
```

```python
memory = OrderList()
memory.add("M")
memory.add("B")
memory.add("Z")
memory.add("A")
memory.add("X")
print(memory.remove())
print(memory)
```
